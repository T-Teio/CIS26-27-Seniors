# 10.2 The Anatomy of a Computer

## Central Processing Unit (CPU)

The part of a computer that executes machine instructions

## Primary Storage

Electronic circuits that can store data as long as they have electric power e.g. RAM/memory installed within a system

## Secondary Storage

Storage that persists without electricity, e.g. a hard disk

## Hard Disk

A device that stores information on rotating platters with magnetic coating (traditional hard drives) Newer Solid State Drives do not have platters, they store data on chips

## Network

An interconnected system of computers and other devices

# 10.8 Python Errors

## Syntax Error

Compile (Translation) error, that is caught before the program runs. in other words, it stops the program from being able to run.

## Exception (Run-time Error)

The syntax is fine (it is legal), so the program runs, however, because of this esception, the program will stop and give you an error

## Logic Error

Type of runtime error. The program runs normally gives you output as it is supposed to. The problem is, if you aren't paying attention, the output is INCORRECT. Ex. print(2 + 2)   OUTPUT: fish

Logic Errors can be difficult to spot. This is why you have to have an understanding of what the output should be.

# 10.9 Algorithm Design

## Algorithm

A recipe for a solution

## Algorithms should be:

### Unambiguous

Percise Intructions at every step. No guesswork or personal opinion.

Example:

"Find the best-looking person."
"Tell me who the best game development studio is."

### Executable

Every step can actually be carried out in practice.

Example: 
"Use next decades real interest rate."

### Terminating

The steps eventually come to an end.
Example:

"Repeat Forever."

### How do we describe an algorithm in pseudocode?

1. Inputs & outputs - What goes in? What do we want to come out?
2. Break it down - split the larger problem into smaller, executable subtasks (problems)
3. Write pseudocode - Order the steps so values exist before they're used
4. Test it - Work a problem by hand with sample values.
